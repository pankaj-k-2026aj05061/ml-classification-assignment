Source note: IBM describes this sample as data for a fictional California telecom company.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
assert (ROOT / 'train_models.py').exists(), 'Open this notebook from the project directory.'
print('Project folder:', ROOT)
!python train_models.py
```
