```python
from pathlib import Path
import os, sys, platform
import pandas as pd
import sklearn

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'train_models.py').exists():
    candidates = [p for p in PROJECT_ROOT.iterdir() if p.is_dir() and (p / 'train_models.py').exists()]
    if candidates:
        PROJECT_ROOT = candidates[0]
if not (PROJECT_ROOT / 'train_models.py').exists():
    raise FileNotFoundError('Run this notebook from the folder containing train_models.py.')
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print('Project folder:', PROJECT_ROOT)
print('Python:', platform.python_version())
print('scikit-learn:', sklearn.__version__)
```

```python
from sklearn.datasets import load_breast_cancer
dataset = load_breast_cancer(as_frame=True)
X = dataset.data
diagnosis = pd.Series(['malignant' if value == 0 else 'benign' for value in dataset.target], name='diagnosis')
print('Dataset: UCI Breast Cancer Wisconsin (Diagnostic)')
print(f'Rows: {len(X)}; predictor features: {X.shape[1]}')
display(diagnosis.value_counts().rename_axis('diagnosis').to_frame('count'))
assert len(X) >= 500 and X.shape[1] >= 12
print('Dataset meets the assignment size requirements.')
```
